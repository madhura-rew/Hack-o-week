# Regression — Week 7 & 8

This Colab notebook covers **Linear Regression, Polynomial Regression, Ridge Regression (L2), and Lasso Regression (L1)**.

## 1. Import Required Libraries

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error
```

## 2. Generate Regression Data

We create a simple dataset where the target is approximately related to the input by **y = 3x + 2**, with some random noise.

```python
np.random.seed(42)

# X = input feature
X_reg = np.random.rand(100, 1) * 10

# y = continuous target
y_reg = 3 * X_reg.squeeze() + 2 + np.random.randn(100) * 2

print("X shape:", X_reg.shape)
print("y shape:", y_reg.shape)
```

## 3. Split Data into Training and Testing Sets

- **80%** data is used for training.
- **20%** data is used for testing.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=42
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

## 4. Linear Regression

Linear Regression fits a straight-line relationship:

**ŷ = wx + b**

It is used when the target is a continuous numerical value.

```python
lin_reg = LinearRegression()

# Train the model
lin_reg.fit(X_train, y_train)

# Make predictions
pred_lin = lin_reg.predict(X_test)

# Evaluate using Mean Squared Error
linear_mse = mean_squared_error(y_test, pred_lin)

print(f"Linear Regression MSE: {linear_mse:.2f}")
print("Weight:", lin_reg.coef_[0])
print("Bias:", lin_reg.intercept_)
```

## 5. Ridge Regression — L2 Regularization

Ridge Regression is Linear Regression with an **L2 penalty**. It discourages very large weights and helps reduce overfitting.

**Loss = MSE + α Σw²**

```python
ridge_reg = Ridge(alpha=1.0)

# Train the model
ridge_reg.fit(X_train, y_train)

# Make predictions
pred_ridge = ridge_reg.predict(X_test)

# Evaluate
ridge_mse = mean_squared_error(y_test, pred_ridge)

print(f"Ridge Regression MSE: {ridge_mse:.2f}")
print("Weight:", ridge_reg.coef_[0])
print("Bias:", ridge_reg.intercept_)
```

## 6. Lasso Regression — L1 Regularization

Lasso Regression is Linear Regression with an **L1 penalty**. It can push some weights exactly to zero, so it can perform feature selection.

**Loss = MSE + α Σ|w|**

```python
lasso_reg = Lasso(alpha=0.1)

# Train the model
lasso_reg.fit(X_train, y_train)

# Make predictions
pred_lasso = lasso_reg.predict(X_test)

# Evaluate
lasso_mse = mean_squared_error(y_test, pred_lasso)

print(f"Lasso Regression MSE: {lasso_mse:.2f}")
print("Weight:", lasso_reg.coef_[0])
print("Bias:", lasso_reg.intercept_)
```

## 7. Polynomial Regression — Degree 2

Polynomial Regression adds powers of the original features, allowing the model to learn a **curved relationship**.

For one feature and degree 2, the model can learn:

**ŷ = w₁x + w₂x² + b**

A `Pipeline` combines the polynomial transformation and Linear Regression.

```python
poly_reg = Pipeline([
    ("poly", PolynomialFeatures(degree=2)),
    ("linear", LinearRegression())
])

# Train the model
poly_reg.fit(X_train, y_train)

# Make predictions
pred_poly = poly_reg.predict(X_test)

# Evaluate
poly_mse = mean_squared_error(y_test, pred_poly)

print(f"Polynomial Regression (degree 2) MSE: {poly_mse:.2f}")
```

## 8. Compare All Regression Models

For regression, **lower MSE is better**.

```python
results = {
    "Linear Regression": linear_mse,
    "Ridge Regression": ridge_mse,
    "Lasso Regression": lasso_mse,
    "Polynomial Regression": poly_mse
}

for model, mse in results.items():
    print(f"{model}: MSE = {mse:.2f}")
```

## 9. Concepts Used

- **Linear Regression:** predicts a continuous value using a straight line.
- **Polynomial Regression:** captures curved relationships by adding powers such as x².
- **Ridge / L2:** reduces large weights and helps control overfitting.
- **Lasso / L1:** can make weights zero and perform feature selection.
- **MSE:** measures average squared prediction error; lower is better.
- **Weights and bias:** parameters learned by the model.
- **Train/test split:** checks how well the trained model performs on unseen data.
