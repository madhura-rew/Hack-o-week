```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import LinearRegression, Ridge, Lasso, LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error, accuracy_score

# --- 1. Generate Dummy Data ---
# Regression Data: X (input), y (continuous target)
np.random.seed(42)
X_reg = np.random.rand(100, 1) * 10
y_reg = 3 * X_reg.squeeze() + 2 + np.random.randn(100) * 2  # y = 3x + 2 + noise

# Classification Data: X (input), y (0 or 1)
X_clf = np.random.rand(100, 2)
# Create a simple separation line for classification
y_clf = (X_clf[:, 0] + X_clf[:, 1] > 1).astype(int)

# Split Data (80% train, 20% test)
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(X_reg, y_reg, test_size=0.2)
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(X_clf, y_clf, test_size=0.2)

print("--- REGRESSION MODELS ---")

# 1. Linear Regression
lin_reg = LinearRegression()
lin_reg.fit(X_train_r, y_train_r)
pred_lin = lin_reg.predict(X_test_r)
print(f"Linear Regression MSE: {mean_squared_error(y_test_r, pred_lin):.2f}")

# 2. Ridge Regression (L2 Regularization)
ridge_reg = Ridge(alpha=1.0)
ridge_reg.fit(X_train_r, y_train_r)
pred_ridge = ridge_reg.predict(X_test_r)
print(f"Ridge Regression MSE: {mean_squared_error(y_test_r, pred_ridge):.2f}")

# 3. Lasso Regression (L1 Regularization)
lasso_reg = Lasso(alpha=0.1)
lasso_reg.fit(X_train_r, y_train_r)
pred_lasso = lasso_reg.predict(X_test_r)
print(f"Lasso Regression MSE: {mean_squared_error(y_test_r, pred_lasso):.2f}")

# 4. Polynomial Regression (Degree 2)
# We use a Pipeline to automatically scale and transform data
poly_reg = Pipeline([
    ('poly', PolynomialFeatures(degree=2)),
    ('linear', LinearRegression())
])
poly_reg.fit(X_train_r, y_train_r)
pred_poly = poly_reg.predict(X_test_r)
print(f"Polynomial Regression (deg 2) MSE: {mean_squared_error(y_test_r, pred_poly):.2f}")

print("\n--- CLASSIFICATION MODELS ---")

# 5. Logistic Regression
# Note: Scaling is important for Logistic Regression and KNN
clf_log = Pipeline([
    ('scaler', StandardScaler()),
    ('log_reg', LogisticRegression())
])
clf_log.fit(X_train_c, y_train_c)
pred_log = clf_log.predict(X_test_c)
print(f"Logistic Regression Accuracy: {accuracy_score(y_test_c, pred_log):.2f}")

# 6. K-Nearest Neighbors (KNN)
clf_knn = Pipeline([
    ('scaler', StandardScaler()),
    ('knn', KNeighborsClassifier(n_neighbors=3))
])
clf_knn.fit(X_train_c, y_train_c)
pred_knn = clf_knn.predict(X_test_c)
print(f"KNN (k=3) Accuracy: {accuracy_score(y_test_c, pred_knn):.2f}")
```
