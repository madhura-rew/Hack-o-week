# Classification — Week 7 & 8

This Colab notebook covers **Logistic Regression and K-Nearest Neighbors (KNN)**.

## 1. Import Required Libraries

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score
```

## 2. Generate Classification Data

We create two input features. The class is **1** when x₁ + x₂ > 1; otherwise the class is **0**.

```python
np.random.seed(42)

# Two input features
X_clf = np.random.rand(100, 2)

# Binary target: 0 or 1
y_clf = (X_clf[:, 0] + X_clf[:, 1] > 1).astype(int)

print("X shape:", X_clf.shape)
print("y shape:", y_clf.shape)
print("Class 0:", np.sum(y_clf == 0))
print("Class 1:", np.sum(y_clf == 1))
```

## 3. Split Data into Training and Testing Sets

- **80%** data is used for training.
- **20%** data is used for testing.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_clf, y_clf, test_size=0.2, random_state=42
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

## 4. Logistic Regression

Despite its name, Logistic Regression is a **classification algorithm**.

It calculates a weighted combination:

**z = wᵀx + b**

Then the sigmoid function converts it into a probability between 0 and 1:

**σ(z) = 1 / (1 + e⁻ᶻ)**

```python
clf_log = Pipeline([
    ("scaler", StandardScaler()),
    ("log_reg", LogisticRegression())
])

# Train the model
clf_log.fit(X_train, y_train)

# Predict classes
pred_log = clf_log.predict(X_test)

# Evaluate
log_accuracy = accuracy_score(y_test, pred_log)

print(f"Logistic Regression Accuracy: {log_accuracy:.2f}")
```

## 5. K-Nearest Neighbors (KNN)

KNN classifies a new point by looking at its nearest training points and taking a majority vote.

Here, **k = 3**, so the 3 nearest neighbors are considered.

Scaling is important because KNN uses distances.

```python
clf_knn = Pipeline([
    ("scaler", StandardScaler()),
    ("knn", KNeighborsClassifier(n_neighbors=3))
])

# Train the model
clf_knn.fit(X_train, y_train)

# Predict classes
pred_knn = clf_knn.predict(X_test)

# Evaluate
knn_accuracy = accuracy_score(y_test, pred_knn)

print(f"KNN (k=3) Accuracy: {knn_accuracy:.2f}")
```

## 6. Compare Classification Models

For classification, **higher accuracy is better**.

```python
results = {
    "Logistic Regression": log_accuracy,
    "KNN (k=3)": knn_accuracy
}

for model, accuracy in results.items():
    print(f"{model}: Accuracy = {accuracy:.2f}")
```

## 7. Concepts Used

- **Classification:** predicts categories/classes such as 0/1.
- **Logistic Regression:** converts a weighted sum into a probability using the sigmoid function, then predicts a class.
- **KNN:** uses nearby training examples and majority voting.
- **StandardScaler:** puts features on a comparable scale; especially important for distance-based KNN.
- **Accuracy:** proportion of correct predictions.
- **Train/test split:** evaluates performance on unseen data.
- **Pipeline:** combines preprocessing and the model into one sequence.
