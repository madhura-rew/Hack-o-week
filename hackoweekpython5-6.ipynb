```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('/content/fifa_world_cup_2026_player_performance.csv')

# Quick look at the data
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
null_counts = df.isnull().sum().sort_values(ascending=False)
null_counts = null_counts[null_counts > 0]
print(null_counts)
```

```python
df.head()
```

### 1. Distribution of Player BMI

This histogram shows the distribution of Body Mass Index (BMI) among players. It helps us understand the general physical build of the players in the dataset. A normal distribution or specific peaks can indicate typical player physiques.

```python
heights=np.array(df['height_cm'])
weights=np.array(df['weight_kg'])

# Calculate BMI and add it to the DataFrame
bmi=(weights)/(heights/100)**2
df['bmi']=bmi

plt.figure(figsize=(10, 6))
sns.histplot(df['bmi'], bins=30, kde=True)
plt.title('Distribution of Player BMI')
plt.xlabel('BMI')
plt.ylabel('Number of Players')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### 2. Distribution of Player Age

This histogram illustrates the age distribution of all players in the dataset. It helps identify the most common age groups among professional footballers in this World Cup data, indicating prime playing years or the presence of young talents and experienced veterans.

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['age'], bins=20, kde=True, color='skyblue')
plt.title('Distribution of Player Age')
plt.xlabel('Age')
plt.ylabel('Number of Players')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### 3. Distribution of Player Market Value (EUR)

This histogram displays the distribution of player market values. It helps in understanding the economic landscape of player valuation within the dataset. The distribution is often skewed, with a few players having very high market values and many having lower values.

```python
plt.figure(figsize=(12, 7))
sns.histplot(df['market_value_eur'], bins=50, kde=True, color='lightcoral')
plt.title('Distribution of Player Market Value (EUR)')
plt.xlabel('Market Value (EUR)')
plt.ylabel('Number of Players')
plt.ticklabel_format(style='plain', axis='x') # Prevent scientific notation on x-axis
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### 4. Average Tournament Rating by Position

This bar chart compares the average tournament rating across different player positions. It can reveal which positions, on average, tend to receive higher ratings, potentially indicating the impact or prominence of certain roles in the game or how performance is perceived for different positions.

```python
avg_rating_by_position = df.groupby('position')['tournament_rating'].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x=avg_rating_by_position.index, y=avg_rating_by_position.values, palette='viridis')
plt.title('Average Tournament Rating by Position')
plt.xlabel('Position')
plt.ylabel('Average Tournament Rating')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
